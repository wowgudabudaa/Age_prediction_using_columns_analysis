## Construct Graphs

```python
# CONSTRUCT GRAPHS 

import os
import numpy as np
import pandas as pd
import torch
from torch_geometric.data import Data
from sklearn.neighbors import NearestNeighbors
```

```python
# Get subject's data (dict: region → matrix)
subject_data = column_data[subj_id]

#Normalize
from sklearn.preprocessing import StandardScaler

# Concatenate all regions for one subject
x_np = np.vstack(list(subject_data.values()))  # shape [n_total_columns, 21]

# Z-score normalization across all features (columns)
scaler = StandardScaler()
x_np = scaler.fit_transform(x_np)
```

```python
# === STEP 1: Build edge_index from K-Nearest Neighbors ===

def build_knn_edge_index(x_np, k=10):
    """
    Builds the edge_index for a graph using KNN based on node features.

    Args:
        x_np (np.ndarray): shape [n_nodes, n_features]
        k (int): number of neighbors per node

    Returns:
        edge_index (torch.LongTensor): shape [2, num_edges]
    """
    knn = NearestNeighbors(n_neighbors=k, metric="euclidean")
    knn.fit(x_np)
    knn_graph = knn.kneighbors_graph(x_np).tocoo()

    edge_index = torch.tensor(
        np.vstack((knn_graph.row, knn_graph.col)),
        dtype=torch.long
    )
    return edge_index
```

```python
# === STEP 2: Build PyG graph from a single subject ===

def build_graph_from_subject(subj_id, column_data, age_dict, k=10):
    """
    Converts one subject's cortical column data into a PyTorch Geometric graph.

    Args:
        subj_id (str): subject ID
        column_data (dict): {subject_id → {region_name → matrix}}
        age_dict (dict): {subject_id → age}
        k (int): number of KNN neighbors

    Returns:
        Data: PyTorch Geometric graph with x, edge_index, y
    """

    # Get all regions from this subject and stack all columns
    subj_regions = column_data[subj_id]
    x_np = np.vstack(list(subj_regions.values()))  # shape [n_columns_total, 22]


    # Convert features to tensor
    x = torch.tensor(x_np, dtype=torch.float32)


    # Build edge_index using KNN over node features
    edge_index = build_knn_edge_index(x_np, k=k)


    # Get subject's age and convert to tensor
    y = torch.tensor([age_dict[subj_id]], dtype=torch.float32)


    # Create PyG Data object
    data = Data(x=x, edge_index=edge_index, y=y)
    data.subject_id = subj_id  # optional: store ID for reference
    return data
```

```python
# === STEP 3: Loop through all subjects and build their graphs ===
graph_data_list = []

for subj_id in column_data:
    if subj_id in age_dict:
        graph = build_graph_from_subject(subj_id, column_data, age_dict, k=10)
        graph_data_list.append(graph)
        print(f" Built graph for subject {subj_id} with {graph.num_nodes} nodes")
    else:
        print(f"[WARNING] No age found for subject {subj_id}")
```

## Define Model GCN

```python
#DEFINE MODEL GCNs
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv, global_mean_pool, BatchNorm

class BrainAgeGNN(torch.nn.Module):
    def __init__(self, input_dim=22):  # ← set to 22 features per node
        super(BrainAgeGNN, self).__init__()

        self.conv1 = GCNConv(input_dim, 64)
        self.bn1 = BatchNorm(64)

        self.conv2 = GCNConv(64, 128)
        self.bn2 = BatchNorm(128)

        self.conv3 = GCNConv(128, 128)
        self.bn3 = BatchNorm(128)

        self.fc = torch.nn.Linear(128, 1)
        self.dropout = torch.nn.Dropout(p=0.5)

    def forward(self, data):
        x, edge_index, batch = data.x, data.edge_index, data.batch

        x = F.relu(self.bn1(self.conv1(x, edge_index)))
        x = F.relu(self.bn2(self.conv2(x, edge_index)))
        x_residual = x
        x = F.relu(self.bn3(self.conv3(x, edge_index)))
        x = x + x_residual

        x = global_mean_pool(x, batch)
        x = self.dropout(x)
        x = self.fc(x)

        return x.view(-1)
```

```python
  #train and eval funcions  
    
from torch.optim import Adam
from torch_geometric.loader import DataLoader  # Usamos el DataLoader de torch_geometric

def train(model, train_loader, optimizer, criterion):
    model.train()
    total_loss = 0
    for data in train_loader:
        data = data.to(device)  # GPU
        optimizer.zero_grad()
        output = model(data).view(-1)
        loss = criterion(output, data.y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    return total_loss / len(train_loader)


def evaluate(model, test_loader, criterion):
    model.eval()
    total_loss = 0
    with torch.no_grad():
        for data in test_loader:
            data = data.to(device)  # GPU
            output = model(data).view(-1)
            loss = criterion(output, data.y)
            total_loss += loss.item()
    return total_loss / len(test_loader)
```

## Training

```python
#Training

import matplotlib.pyplot as plt
from sklearn.model_selection import StratifiedKFold
import numpy as np
import pandas as pd

# Training settings
epochs = 300
patience = 40
k = 7  # Number of folds
batch_size = 6

# === Initialize loss and tracking ===
all_train_losses = []
all_test_losses = []
all_early_stopping_epochs = []

# === Extract subject IDs from graph list ===
graph_subject_ids = [data.subject_id for data in graph_data_list]

# === Get matching ages in the same order ===
ages = np.array([age_dict[subj_id] for subj_id in graph_subject_ids])

# === Create age bins for stratification ===
age_bins = pd.qcut(ages, q=5, labels=False)  # 5 quantiles
```

```python
# Stratified split by age bins
skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)
repeats_per_fold = 10
```

```python
# === Build age vector and age bins ===
graph_subject_ids = [data.subject_id for data in graph_data_list]
ages = np.array([age_dict[subj_id] for subj_id in graph_subject_ids])
age_bins = pd.qcut(ages, q=5, labels=False)
```

```python

for fold in range(k):
    train_idx = np.loadtxt(f"../set_split/train_indices_fold_{fold}.csv", delimiter=",").astype(np.int64)
    test_idx = np.loadtxt(f"../set_split/test_indices_fold_{fold}.csv", delimiter=",").astype(np.int64)
    print(f"\n=== Fold {fold+1}/{k} ===")

    train_data = [graph_data_list[i] for i in train_idx]
    test_data = [graph_data_list[i] for i in test_idx]

    fold_train_losses = []
    fold_test_losses = []

    for repeat in range(repeats_per_fold):
        print(f'  > Repeat {repeat+1}/{repeats_per_fold}')

        early_stop_epoch = None
        seed_everything(42 + repeat)

        train_loader = DataLoader(train_data, batch_size=batch_size, shuffle=True)
        test_loader = DataLoader(test_data, batch_size=batch_size, shuffle=False)

        model = BrainAgeGNN(input_dim=22).to(device)

        optimizer = AdamW(model.parameters(), lr=0.002, weight_decay=1e-4)
        scheduler = StepLR(optimizer, step_size=20, gamma=0.5)
        criterion = torch.nn.SmoothL1Loss(beta=1)

        best_loss = float('inf')
        patience_counter = 0

        train_losses = []
        test_losses = []

        for epoch in range(epochs):
            train_loss = train(model, train_loader, optimizer, criterion)
            test_loss = evaluate(model, test_loader, criterion)

            train_losses.append(train_loss)
            test_losses.append(test_loss)

            if test_loss < best_loss:
                best_loss = test_loss
                patience_counter = 0
                torch.save(model.state_dict(), f"column_md_model_fold_{fold+1}_rep_{repeat+1}_ref.pt")
            else:
                patience_counter += 1
                if patience_counter >= patience:
                    early_stop_epoch = epoch + 1
                    print(f"    Early stopping triggered at epoch {early_stop_epoch}.")
                    break

            scheduler.step()

        if early_stop_epoch is None:
            early_stop_epoch = epochs

        all_early_stopping_epochs.append((fold + 1, repeat + 1, early_stop_epoch))

        fold_train_losses.append(train_losses)
        fold_test_losses.append(test_losses)

    all_train_losses.append(fold_train_losses)
    all_test_losses.append(fold_test_losses)
```

```python
#######################  LEARNING CURVE (MEAN ± STD)  ########################
# Compute mean and std per epoch across all folds and repetitions
avg_train = []
avg_test = []

for epoch in range(epochs):
    epoch_train = []
    epoch_test = []
    for fold in range(k):
        for rep in range(repeats_per_fold):
            if epoch < len(all_train_losses[fold][rep]):
                epoch_train.append(all_train_losses[fold][rep][epoch])
                epoch_test.append(all_test_losses[fold][rep][epoch])
    avg_train.append((np.mean(epoch_train), np.std(epoch_train)))
    avg_test.append((np.mean(epoch_test), np.std(epoch_test)))

```

```python
# Unpack values
train_mean, train_std = zip(*avg_train)
test_mean, test_std = zip(*avg_test)
```

```python
# Plot mean ± std
plt.figure(figsize=(10, 6))

plt.plot(train_mean, label="Train Mean", color="blue")
plt.fill_between(range(epochs), np.array(train_mean) - np.array(train_std),
                 np.array(train_mean) + np.array(train_std), color="blue", alpha=0.3)

plt.plot(test_mean, label="Test Mean", color="orange")
plt.fill_between(range(epochs), np.array(test_mean) - np.array(test_std),
                 np.array(test_mean) + np.array(test_std), color="orange", alpha=0.3)

plt.xlabel("Epoch")
plt.ylabel("Smooth L1 Loss")
plt.title("Learning Curve (Mean ± Std – GCN on Column-Based Graphs)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
##################  PREDICTION & METRIC ANALYSIS (COLUMN-BASED GRAPHS)  ##################

from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error
import matplotlib.pyplot as plt

# Ignore future warnings
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)

# === Initialize storage ===
fold_mae_list = []
fold_r2_list = []
fold_rmse_list = []

all_subject_ids = []
all_y_true = []
all_y_pred = []

# Create a new folder to save the y_true, y_pred, and subject_ids in each fold and repeat
if not os.path.exists("column_md_results"):
    os.makedirs("column_md_results")

for fold in range(k):
    train_idx = np.loadtxt(f"../set_split/train_indices_fold_{fold}.csv", delimiter=",", dtype=int)
    test_idx = np.loadtxt(f"../set_split/test_indices_fold_{fold}.csv", delimiter=",", dtype=int)
    print(f"\n=== Evaluating Fold {fold+1}/{k} ===")

    test_data = [graph_data_list[i] for i in test_idx]
    test_loader = DataLoader(test_data, batch_size=batch_size, shuffle=False)
    train_data = [graph_data_list[i] for i in train_idx]
    train_loader = DataLoader(train_data, batch_size=batch_size, shuffle=False)

    repeat_maes = []
    repeat_r2s = []
    repeat_rmses = []

    for rep in range(repeats_per_fold):
        print(f"  > Repeat {rep+1}/{repeats_per_fold}")

        model = BrainAgeGNN(input_dim=22).to(device)  # ← Your model

        model.load_state_dict(torch.load(f"column_md_model_fold_{fold+1}_rep_{rep+1}_ref.pt"))  # Load saved model
        model.eval()

        subject_id_repeat = []
        y_true_repeat = []
        y_pred_repeat = []

        subject_id_repeat_train = []
        y_true_repeat_train = []
        y_pred_repeat_train = []

        with torch.no_grad():
            for data in test_loader:
                data = data.to(device)
                output = model(data).view(-1)
                subject_id_repeat.extend(data.subject_id)
                y_pred_repeat.extend(output.cpu().tolist())
                y_true_repeat.extend(data.y.cpu().tolist())
            # Save predictions and true values to files
            test_results_df = pd.DataFrame({
                'subject_id': subject_id_repeat,
                'y_true': y_true_repeat,
                'y_pred': y_pred_repeat
            })
            test_results_df.to_csv(f"column_md_results/fold_{fold+1}_rep_{rep+1}_test_results_ref.csv", index=False)

            for data in train_loader:
                data = data.to(device)
                output = model(data).view(-1)
                subject_id_repeat_train.extend(data.subject_id)
                y_pred_repeat_train.extend(output.cpu().tolist())
                y_true_repeat_train.extend(data.y.cpu().tolist())
            # Save predictions and true values for training data
            train_results_df = pd.DataFrame({
                'subject_id': subject_id_repeat_train,
                'y_true': y_true_repeat_train,
                'y_pred': y_pred_repeat_train
            })
            train_results_df.to_csv(f"column_md_results/fold_{fold+1}_rep_{rep+1}_train_results_ref.csv", index=False)

        # Compute metrics
        mae = mean_absolute_error(y_true_repeat, y_pred_repeat)
        r2 = r2_score(y_true_repeat, y_pred_repeat)
        rmse = np.sqrt(mean_squared_error(y_true_repeat, y_pred_repeat))

        repeat_maes.append(mae)
        repeat_r2s.append(r2)
        repeat_rmses.append(rmse)

        all_y_true.extend(y_true_repeat)
        all_y_pred.extend(y_pred_repeat)
        all_subject_ids.extend(subject_id_repeat)

    fold_mae_list.append(repeat_maes)
    fold_r2_list.append(repeat_r2s)
    fold_rmse_list.append(repeat_rmses)

    print(f">> Fold {fold+1} | MAE: {np.mean(repeat_maes):.2f} ± {np.std(repeat_maes):.2f} | "
          f"R²: {np.mean(repeat_r2s):.2f} ± {np.std(repeat_r2s):.2f} | "
          f"RMSE: {np.mean(repeat_rmses):.2f} ± {np.std(repeat_rmses):.2f}")

```

```python
# === DIRECT SHAP FOR THE GCN (per-fold, per-repeat, GradientExplainer) ===
import joblib
import torch
import torch.nn as nn
from torch_geometric.data import Data
#from torch_scatter import scatter_mean  # if needed, but we use global_mean_pool from PyG directly
from torch_geometric.nn import global_mean_pool
import shap

# Config
n_nodes = len(region_keys)           # should be 68
n_feat = 22                          # 22 features per node
flat_dim = n_nodes * n_feat          # 1496
bg_max = 50                          # number of background samples from training per fold/rep
shap_dir = "shap_gcn_results"
os.makedirs(shap_dir, exist_ok=True)

# Build flattened feature matrix for all subjects in the same order as graph_data_list
X_all = np.stack([g.x.numpy().reshape(-1).astype(np.float32) for g in graph_data_list], axis=0)  # (n_subjects, 1496)
y_all = np.array([age_dict[g.subject_id] for g in graph_data_list], dtype=np.float32)

# Build a mapping from flattened bytes -> index so wrapper can retrieve the original edge_index for each sample
flat_to_idx = {X_all[i].tobytes(): i for i in range(X_all.shape[0])}
edge_index_list = [g.edge_index for g in graph_data_list]  # list of torch tensors

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

class GNNTorchWrapper(nn.Module):
    """
    Wraps your BrainAgeGNN so shap.GradientExplainer can call it with a torch.Tensor input
    of shape (B, n_nodes * n_feat). The wrapper rebuilds a batched PyG graph inside forward.
    """
    def __init__(self, gnn_model, edge_index_list, n_nodes, n_feat, k_for_knn=None):
        super().__init__()
        self.gnn = gnn_model
        self.edge_index_list = edge_index_list  # list of torch.LongTensor edge_index per subject
        self.n_nodes = n_nodes
        self.n_feat = n_feat
        self.k_for_knn = k_for_knn  # optional fallback if edge_index missing
        # ensure model on same device
        self.to(device)

    def forward(self, x_flat):
        """
        x_flat: Tensor shape (B, n_nodes * n_feat)  dtype float32, device same as self.gnn
        returns: Tensor shape (B,) predictions
        """
        # Convert numpy array to torch tensor if needed
        if isinstance(x_flat, np.ndarray):
            x_flat = torch.from_numpy(x_flat).float().to(device)
        assert x_flat.dim() == 2
        B = x_flat.size(0)
        xs = []
        batch_ids = []
        edge_index_blocks = []
        total_nodes = 0

        # Build concatenated node-features and build batched edge_index with offsets
        for i in range(B):
            row = x_flat[i]                          # (n_nodes * n_feat,)
            # Ensure row is a torch tensor on the correct device
            if not isinstance(row, torch.Tensor):
                row = torch.from_numpy(np.array(row)).float()
            # Only call .to(device) if row is a torch tensor
            if isinstance(row, torch.Tensor):
                row = row.to(device)
            x_i = row.view(self.n_nodes, self.n_feat)   # (n_nodes, n_feat)
            xs.append(x_i)

            # get edge_index for this sample if available
            if i < len(self.edge_index_list) and self.edge_index_list[i] is not None:
                ei = self.edge_index_list[i]
                if isinstance(ei, np.ndarray):
                    ei = torch.from_numpy(ei).long()
                ei = ei.to(device)
            else:
                # fallback: build using KNN from features (you must have build_knn_edge_index that returns torch.LongTensor)
                if self.k_for_knn is None:
                    # if no fallback, use first edge_index as proxy
                    ei = self.edge_index_list[0]
                    if isinstance(ei, np.ndarray):
                        ei = torch.from_numpy(ei).long()
                    ei = ei.to(device)
                else:
                    x_np = x_i.cpu().numpy()
                    ei = build_knn_edge_index(x_np, k=self.k_for_knn)
                    if isinstance(ei, np.ndarray):
                        ei = torch.from_numpy(ei).long()
                    ei = ei.to(device)

            # offset indices by total_nodes and append
            edge_index_blocks.append(ei + total_nodes)
            # increment total nodes
            total_nodes += x_i.size(0)
            # accumulate batch ids
            batch_ids.append(torch.full((x_i.size(0),), i, dtype=torch.long, device=device))

        x_cat = torch.cat(xs, dim=0)            # (B * n_nodes, n_feat)
        batch_tensor = torch.cat(batch_ids, dim=0)  # (B * n_nodes,)

        # combine edge_index blocks into single tensor
        edge_index = torch.cat(edge_index_blocks, dim=1)  # (2, total_edges)

        # Build temporary Data object and forward through GNN
        data = Data(x=x_cat, edge_index=edge_index, batch=batch_tensor)
        data = data.to(device)
        out = self.gnn(data)    # shape (B,) or (B,)
        # Ensure we return shape (B, num_outputs). Here num_outputs = 1
        if out.dim() == 1:
            out = out.view(-1, 1)   # becomes (B,1)
        else:
            out = out.view(out.size(0), -1)
        return out

# Storage for aggregation
model_region_shaps = []   # list of (n_regions,) arrays (mean abs SHAP per region) per model
model_meta = []           # list of (fold,rep) pairs

# Loop through folds and repeats (mirror your evaluation)
for fold in range(k):
    # load precomputed train/test indices (you already save them)
    train_idx = np.loadtxt(f"../set_split/train_indices_fold_{fold}.csv", delimiter=",").astype(np.int64)
    test_idx  = np.loadtxt(f"../set_split/test_indices_fold_{fold}.csv", delimiter=",").astype(np.int64)
    print(f"\n[SHAP] Processing fold {fold+1}/{k}")
    X_train = X_all[train_idx]
    X_test  = X_all[test_idx]

    for rep in range(repeats_per_fold):
        print(f"[SHAP] computing fold {fold+1}/{k} rep {rep+1}/{repeats_per_fold}")

        # load trained model for this fold/rep (you saved these during training)
        model = BrainAgeGNN(input_dim=n_feat).to(device)
        model.load_state_dict(torch.load(f"column_md_model_fold_{fold+1}_rep_{rep+1}_ref.pt", map_location=device))
        model.eval()

        # choose background samples from training set (no leakage)
        rng = np.random.default_rng(seed=fold*100 + rep)
        bg_k = min(bg_max, X_train.shape[0])
        bg_idx_local = rng.choice(X_train.shape[0], size=bg_k, replace=False)
        background = X_train[bg_idx_local]  # numpy array (bg_k, flat_dim)

        # Make callable (captures model & edge_index_list)
        callable_model = GNNTorchWrapper(model, edge_index_list, n_nodes, n_feat)

        # Create GradientExplainer (uses autograd)
        explainer = shap.GradientExplainer(callable_model, background)

        # --- convert background into torch tensor (device-aware) and enable grad
        background_np = X_train[bg_idx_local]               # numpy (bg_k, flat_dim)
        background_t = torch.from_numpy(background_np).float().to(device)
        background_t.requires_grad_(True)                   # IMPORTANT for GradientExplainer

        # Make callable (captures model & edge_index_list)
        callable_model = GNNTorchWrapper(model, edge_index_list, n_nodes, n_feat)
        callable_model.to(device)
        callable_model.eval()

        # Create GradientExplainer (expects torch tensors / torch.nn.Module)
        explainer = shap.GradientExplainer(callable_model, background_t)

        # Prepare X_test as torch tensor on device (you can pass cpu tensor too)
        X_test_t = torch.from_numpy(X_test).float().to(device)

        # Compute shap values. For large test sets compute in chunks to avoid OOM / long runtimes.
        # We'll explain in chunks of size `explain_batch` to be memory friendly.
        explain_batch = 32
        shap_vals_list = []
        for i in range(0, X_test_t.size(0), explain_batch):
            X_chunk = X_test_t[i:i+explain_batch]
            # shap returns numpy arrays usually; with torch background it returns tensors or arrays depending on SHAP version
            sv = explainer.shap_values(X_chunk)   # may return a tensor or list
            # normalize output to numpy array
            if isinstance(sv, list):
                sv = sv[0]
            if isinstance(sv, torch.Tensor):
                sv = sv.detach().cpu().numpy()
            shap_vals_list.append(sv)

        shap_vals = np.vstack(shap_vals_list)   # (n_test, flat_dim)

        # Some shap versions wrap result in list for multiclass; ensure numpy:
        if isinstance(shap_vals, list):
            shap_vals = shap_vals[0]

        # Save raw shap array for this fold/rep
        np.save(os.path.join(shap_dir, f"shap_fold{fold+1}_rep{rep+1}.npy"), shap_vals)
        joblib.dump(model.state_dict(), os.path.join(shap_dir, f"gcn_state_fold{fold+1}_rep{rep+1}.joblib"))

        # Aggregate to region-level mean absolute SHAP for this model
        shap_abs = np.abs(shap_vals)  # (n_test, flat_dim)
        mean_abs_by_feature = shap_abs.mean(axis=0)  # (flat_dim,)
        region_shap = mean_abs_by_feature.reshape(n_nodes, n_feat).mean(axis=1)  # (n_regions,)
        model_region_shaps.append(region_shap)
        model_meta.append((fold, rep))

        # Save per-region CSV for this model
        pd.DataFrame({'region': region_keys, 'mean_abs_shap': region_shap}).to_csv(
            os.path.join(shap_dir, f"region_shap_fold{fold+1}_rep{rep+1}.csv"), index=False
        )

# Final aggregation across models
model_region_shaps = np.stack(model_region_shaps, axis=0)  # (n_models, n_regions)
region_mean = model_region_shaps.mean(axis=0)
region_std  = model_region_shaps.std(axis=0)
df_summary = pd.DataFrame({'region': region_keys, 'mean_abs_shap': region_mean, 'std_abs_shap': region_std})
df_summary = df_summary.sort_values('mean_abs_shap', ascending=False)
df_summary.to_csv(os.path.join(shap_dir, "region_shap_summary_all_folds_reps.csv"), index=False)
print("[SHAP] done — summary saved at", os.path.join(shap_dir, "region_shap_summary_all_folds_reps.csv"))

# Quick plot of top regions
topk = 15
top = df_summary.head(topk).iloc[::-1]
plt.figure(figsize=(8,6))
plt.barh(top['region'], top['mean_abs_shap'], xerr=top['std_abs_shap'], ecolor='gray', capsize=3)
plt.xlabel("Mean |SHAP| (aggregated)")
plt.title(f"Top {topk} regions by mean |SHAP| (GNN, aggregated across folds/repeats)")
plt.tight_layout()
plt.show()
```
